This project outlines where the money is being spent and in what categories are spending the most money

```python
import pandas as pd

df = pd.read_csv("transactions.csv")
print(df.head())
print("Total spent:", df["Amount"].sum())
```

```python
categories = {
    "CHIPOTLE": "Food",
    "STARBUCKS": "Food",
    "WALMART": "Groceries",
    "SHELL": "Gas",
    "UBER": "Transport",
    "Spotify" : "Subscription",
    "DoorDash" : "Food",
    "Taco Bell" : "Food",
    "Planet Fitness" : "Subscription",
    "Netflix" : "Subscription",
    "Apple" : "Subscription",
    "Verizon" : "Subscription",
    "Dominos" : "Food",
    "Amazon" : "Shopping",
    "Target" : "Shopping",
    "CVS" : "Shopping",
    "Campus BookStore" : "School",
    "Theatres" : "Entertainment",
    "Games" : "Entertainment"


}

def categorize(description):
    for keyword, category in categories.items():
        if keyword.upper() in description.upper():
            return category
    return "Other"

df["Category"] = df["Description"].apply(categorize)
print(df.groupby("Category")["Amount"].sum())
```

```python
import matplotlib.pyplot as plt

totals = df.groupby("Category")["Amount"].sum().sort_values()
totals.plot(kind="barh", title="September spending by category")
plt.xlabel("Dollars")
plt.show()

print((totals / totals.sum() * 100).round(1))
```
